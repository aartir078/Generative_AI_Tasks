```python
import random
import re
from collections import defaultdict

# Sample training text
text = """
Artificial intelligence is changing the world.
Artificial intelligence helps people learn.
Machine learning helps computers learn from data.
Python is useful for machine learning.
Artificial intelligence can generate text.
Python is popular for data science.
Machine learning is a part of artificial intelligence.
"""

# Convert text into words
words = re.findall(r"\b\w+\b|[.!?]", text.lower())

# Build the Markov chain
model = defaultdict(list)

for i in range(len(words) - 2):
    current = (words[i], words[i + 1])
    next_word = words[i + 2]
    model[current].append(next_word)

# Generate text using the learned transitions
def generate_text(start_words, length=25):
    current = tuple(start_words.lower().split())
    result = list(current)

    for _ in range(length):
        choices = model.get(current)

        if not choices:
            break

        next_word = random.choice(choices)
        result.append(next_word)
        current = (current[1], next_word)

    return " ".join(result)

print("Generated Text:")
print(generate_text("artificial intelligence"))
```
